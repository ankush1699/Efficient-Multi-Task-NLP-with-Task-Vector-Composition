```python
import torch
from transformers import AutoModelForSequenceClassification
from torch.nn.utils import parameters_to_vector
```

```python
base_model = AutoModelForSequenceClassification.from_pretrained("base_model")
taskA_model = AutoModelForSequenceClassification.from_pretrained("model_taskA_sst2")
```

```python
for param in base_model.parameters():
    param.requires_grad = False

for param in taskA_model.parameters():
    param.requires_grad = False
```

```python
theta_0 = parameters_to_vector(base_model.parameters()).detach()
theta_A = parameters_to_vector(taskA_model.parameters()).detach()
```

```python
delta_A = theta_A - theta_0
```

```python
torch.save(delta_A, "delta_taskA_sst2.pt")
```
